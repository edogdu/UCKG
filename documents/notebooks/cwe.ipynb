```python
from neo4j import GraphDatabase
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

uri = "bolt://localhost:7687"
username = "neo4j"
password = "abcd90909090"

driver = GraphDatabase.driver(uri, auth=(username, password))

def run_query(query, parameters=None):
    with driver.session() as session:
        result = session.run(query, parameters)
        return [record.data() for record in result]
```

```python
# The number of UcoCWE nodes
query = """
MATCH (n:UcoCWE) RETURN count(n) AS UcoCWE
"""
params = {}

records = run_query(query, parameters=params)

df = pd.DataFrame(records)

df
```

```python
# The properties of UcoCWE nodes
query = """
MATCH (n:UcoCWE)
UNWIND keys(n) AS key
RETURN DISTINCT key
"""
params = {}

records = run_query(query, parameters=params)

df = pd.DataFrame(records)

df
```

```python
# The number of incoming relationships to UcoCWE nodes
query = """
MATCH (n:UcoCWE)<-[r]-(m)
RETURN DISTINCT type(r) AS incomingRelationship, count(r)
"""
params = {}

records = run_query(query, parameters=params)

df = pd.DataFrame(records)

df
```

```python
# The number of outgoing relationships from UcoCWE nodes
query = """
MATCH (n:UcoCWE)-[r]->()
RETURN DISTINCT type(r) AS outgoingRelationship, count(r)
"""
params = {}

records = run_query(query, parameters=params)

df = pd.DataFrame(records)

df
```

```python
# Node labels that are connected to UcoCWE nodes in a hop
query = """
MATCH (n:UcoCWE)-[r]-(m)
UNWIND labels(m) AS label
RETURN DISTINCT label AS targetNodeLabels
"""
params = {}

records = run_query(query, parameters=params)

df = pd.DataFrame(records)

df
```

```python
# The number of UcoExploitTarget nodes connected to UcoCWE nodes
# and the relationship type between them
query = """
MATCH (n:UcoCWE)<-[r]-(m:UcoExploitTarget)
RETURN DISTINCT count(m) AS numOfUcoExploitTarget, type(r) AS RelationshipType
"""
params = {}

records = run_query(query, parameters=params)

df = pd.DataFrame(records)

df
```

```python
# The number of UcoexCAPEC nodes connected to UcoCWE nodes
# and the relationship type between them
query = """
MATCH (n:UcoCWE)<-[r]-(m:UcoexCAPEC)
RETURN DISTINCT count(m) AS numOfUcoexCAPEC, type(r) AS RelationshipType
"""
params = {}

records = run_query(query, parameters=params)

df = pd.DataFrame(records)

df
```

```python
# The number of UcoexObservedExample(connected to UcoCVE) nodes connected to UcoCWE nodes
# and the relationship type between them
query = """
MATCH (n:UcoCWE)-[r]->(m:UcoexObservedExample)
RETURN DISTINCT count(m) AS numOfUcoexObservedExample, type(r) AS RelationshipType
"""
params = {}

records = run_query(query, parameters=params)

df = pd.DataFrame(records)

df
```

```python
# The number of UcoCWE nodes and UcoexCPE nodes connected to each other in 3 hops
query = """
MATCH p = (n:UcoCWE)-[*1..3]-(m:UcoexCPE)
RETURN count(DISTINCT n) AS numOfCWE, count(DISTINCT m) AS numOfCPE
"""
params = {}

records = run_query(query, parameters=params)

df = pd.DataFrame(records)

df
```

```python
# The number of UcoCWE nodes and UcoexCAPEC nodes connected to each other
query = """
MATCH p = (n:UcoCWE)-[r]-(m:UcoexCAPEC)
RETURN count(DISTINCT n) AS numOfCWE, count(DISTINCT m) AS numOfCAPEC
"""
params = {}

records = run_query(query, parameters=params)

df = pd.DataFrame(records)

df
```
